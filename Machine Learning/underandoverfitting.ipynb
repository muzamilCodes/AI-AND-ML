```python
                                                                  # Underfitting and Overfitting
# 1. Underfitting

# Underfitting happens when a Machine Learning model is too simple to learn the important patterns in the data. Because the model does not learn enough, it performs poorly on both training and test data.

# Simple meaning:
# 👉 Model ne data se enough nahi seekha.
# Characteristics of Underfitting
# - ❌ Poor performance on training data
# - ❌ Poor performance on test data
# - 📉 High training error
# - 📉 High test error
# - 🧠 Model is too simple
# - 🔍 Important patterns are not captured
# - ⚙️ Usually caused by:
#   - Model being too simple
#   - Too few features
#   - Insufficient training
#   - Excessive regularization
# Example:
# Using a straight line to predict data that actually follows a complex curve.
```

```python
# 2. Overfitting
# Overfitting happens when a Machine Learning model learns the training data too well, including noise and random fluctuations. It then performs very well on training data but poorly on new/unseen test data.

# Simple meaning:
# 👉 Model ne training data ko yaad kar liya, lekin new data par acha kaam nahi karta.
# Characteristics of Overfitting
# - ✅ Very high performance on training data
# - ❌ Poor performance on test data
# - 📉 Very low training error
# - 📈 High test/validation error
# - 🧠 Model is too complex
# - 🔍 Learns noise and unnecessary details
# - ⚠️ Poor generalization to unseen data
# - ⚙️ Usually caused by:
#   - Model being too complex
#   - Too many features
#   - Too little training data
#   - Training for too long
#   - Lack of regularization
# Example:
# A very high-degree polynomial that passes through almost every training point but makes bad predictions for new points.
```

```python
# Bias in Machine Learning
# Bias is the error caused when a Machine Learning model makes too many assumptions about the data. A high-bias model is usually too simple and cannot capture the important patterns in the data.
# Characteristics of Bias:
# - High bias means the model is too simple.
# - The model does not learn enough from the training data.
# - It gives poor performance on training data.
# - It also gives poor performance on test data.
# - It usually causes underfitting.
# - The model makes strong assumptions about the data.
# - It has high training error and high testing error.
# Example:
# Suppose the actual relationship between Experience and Salary is curved, but we use a simple linear model:
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4, 5, 6, 7, 8])

y_actual = np.array([15, 25, 40, 55, 70, 90, 105, 130])

actual_model = y_actual

predicted_model = np.array([50, 50, 50, 50, 50, 50, 50, 50])

plt.figure(figsize=(12, 6), dpi=500)

plt.plot(x, actual_model, marker='o', label='Actual Model')
plt.plot(x, predicted_model, marker='o', label='Predicted Model')

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Actual vs Predicted Model")

plt.legend()
plt.grid(True)

plt.show()
```
