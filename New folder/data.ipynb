```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('ggplot')

import warnings
warnings.filterwarnings('ignore')

%matplotlib inline

print("Everything is working!")
```

```python
data=pd.read_csv("CO2_Emissions_Canada.csv")
data.head(5)
```

```python
plt.figure(figsize=(12,6),dpi= 500)
sns.scatterplot(x="Engine Size(L)",y="CO2 Emissions(g/km)",data=data)
plt.show()
```

```python

plt.figure(figsize=(12, 6), dpi=500)

plt.subplot(1, 2, 1)
sns.barplot(x="Cylinders", y="CO2 Emissions(g/km)", data=data)
plt.title("Cylinders vs CO2 Emissions (Bar Plot)")

plt.subplot(1, 2, 2)
sns.boxplot(x="Cylinders", y="CO2 Emissions(g/km)", data=data)
plt.title("Cylinders vs CO2 Emissions (Box Plot)")

plt.tight_layout()
plt.show()


```
